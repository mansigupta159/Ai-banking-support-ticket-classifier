```python
X = df["case_description"]
y = df["category"]

print("Input features:", X.shape)
print("Target values:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", LogisticRegression(max_iter=1000))
])

print("Machine Learning model created successfully.")
```

```python
model.fit(X_train, y_train)

print("Model training completed successfully.")
```

```python
predictions = model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)

print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python
print(classification_report(
    y_test,
    predictions,
    zero_division=0
))
```

```python
CONFIDENCE_THRESHOLD = 50
```

```python
def get_priority(category, case_description):
    case_description = case_description.lower()

    high_priority_keywords = [
        "payment failed",
        "transaction failed",
        "cif already exists",
        "cannot login",
        "account creation failed"
    ]

    for keyword in high_priority_keywords:
        if keyword in case_description:
            return "High"

    if category in ["Payment", "Account Creation"]:
        return "Medium"

    if category == "Out of Scope":
        return "N/A"

    return "Low"
```

```python
def analyze_case(case_text):

    probabilities = model.predict_proba([case_text])

    confidence = np.max(probabilities) * 100

    category = model.predict([case_text])[0]

    # Out-of-scope handling
    if confidence < CONFIDENCE_THRESHOLD:
        return {
            "Case": case_text,
            "Category": "Unknown / Out of Scope",
            "Priority": "N/A",
            "Confidence": round(confidence, 2),
            "Suggested Action": "Route the case for manual review."
        }

    # If model itself predicts Out of Scope
    if category == "Out of Scope":
        return {
            "Case": case_text,
            "Category": "Out of Scope",
            "Priority": "N/A",
            "Confidence": round(confidence, 2),
            "Suggested Action": "This issue is outside the supported banking categories."
        }

    priority = get_priority(category, case_text)

    suggested_actions = {
        "Authentication":
            "Check login, password and authentication logs.",

        "Account Creation":
            "Check customer/CIF mapping and account creation flow.",

        "Payment":
            "Check transaction status and payment processing logs.",

        "KYC":
            "Check KYC documents and verification status.",

        "Customer Profile":
            "Check customer profile data and update transaction."
    }

    return {
        "Case": case_text,
        "Category": category,
        "Priority": priority,
        "Confidence": round(confidence, 2),
        "Suggested Action": suggested_actions.get(
            category,
            "Perform manual investigation."
        )
    }
```

```python
test_cases = [
    "Customer cannot login to banking application",
    "CIF already exists while creating account",
    "Customer payment transaction failed",
    "Customer KYC document was rejected",
    "Customer wants to change mobile number",
    "My laptop keyboard is not working",
    "My printer is not working"
]

for case in test_cases:

    result = analyze_case(case)

    print("--------------------------------------------")
    print("Case:", result["Case"])
    print("Category:", result["Category"])
    print("Priority:", result["Priority"])
    print("Confidence:", result["Confidence"], "%")
    print("Suggested Action:", result["Suggested Action"])
```
